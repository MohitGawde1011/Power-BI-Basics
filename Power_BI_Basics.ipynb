1. What is Power BI and why is it used in businesses?

-  Power BI is a business intelligence and data visualization tool developed by Microsoft. It is used to connect, clean, transform, analyze, and visualize data.

Businesses use Power BI to:

Create interactive dashboards and reports.
Analyze sales, finance, customers, and other business data.
Identify trends and patterns.
Monitor KPIs and business performance.
Make faster and data-driven decisions.
Share reports and insights with teams.


2. Name and explain the three main components of Power BI.

-  The three main components are:

Power BI Desktop
A Windows application used to import data, clean and transform it, create data models, and design reports and visualizations.
Power BI Service
A cloud-based platform used to publish, share, collaborate on, and access Power BI reports and dashboards online.
Power BI Mobile
Mobile applications for Android and iOS that allow users to view and interact with dashboards and reports on smartphones and tablets.


3. Explain the Power BI workflow.

-  The typical Power BI workflow is:

1. Connect → 2. Transform → 3. Model → 4. Visualize → 5. Publish → 6. Share

1. Connect

Connect Power BI to data sources such as Excel, CSV, SQL Server, or web sources.

2. Transform

Use Power Query to clean and transform the data by removing duplicates, handling missing values, changing data types, etc.

3. Model

Create relationships between tables and build a suitable data model.

4. Visualize

Create charts, tables, cards, maps, and other visualizations in Report View.

5. Publish

Publish the completed report from Power BI Desktop to the Power BI Service.

6. Share

Share dashboards and reports with authorized users or teams.

4. List any four data cleaning tasks that can be performed in Power Query.

-  Four common data cleaning tasks are:

Remove duplicates – Removes repeated records.
Handle missing/null values – Replace, remove, or identify missing values.
Change data types – Convert columns to appropriate types such as Text, Number, Date, etc.
Split columns – Divide one column into multiple columns, such as splitting an Owner Name into First Name and Last Name.

Other tasks include removing unnecessary columns, trimming spaces, replacing values, filtering rows, and changing column names.


6. Define Data View, Report View, and Model View.

-  | View            | Definition                                               | Purpose                                                                |
| --------------- | -------------------------------------------------------- | ---------------------------------------------------------------------- |
| **Data View**   | Displays the actual rows and columns of the loaded data. | Used to inspect and analyze data and create calculated columns.        |
| **Report View** | The main area for creating visual reports.               | Used to create charts, tables, cards, slicers, and dashboards.         |
| **Model View**  | Displays tables and relationships between them.          | Used to create and manage relationships and understand the data model. |


1. Data View

Data View allows you to see the actual data in table format.

2. Report View

Report View is used to build interactive reports using visualizations such as bar charts, pie charts, maps, and KPI cards.

3. Model View

Model View shows how different tables are connected through relationships. It is especially useful when working with multiple tables.

7. Discuss the different data sources that Power BI supports.

-   Power BI supports many different types of data sources, including:

1. Excel

Power BI can import tables and worksheets from Excel files.

2. CSV/Text Files

CSV files are commonly used for structured tabular data.

3. Databases

Power BI supports databases such as:

SQL Server
MySQL
Oracle
PostgreSQL
Microsoft Access
4. Web

Power BI can retrieve data from web pages and web-based sources.

5. Cloud Services

It can connect to services such as SharePoint, Azure, and other cloud platforms.

6. APIs

Power BI can connect to APIs to retrieve data from external applications and services.

7. Other Sources

It also supports sources such as JSON, XML, PDF files, and various online services.
